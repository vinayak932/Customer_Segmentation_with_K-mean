```python
import seaborn as sns
columns = [
    "Age",
    "Annual Income (k$)",
    "Spending Score (1-100)"
]
for col in columns:
  fig, axes = plt.subplots(1,2, figsize=(10,3))
  sns.histplot(data=df, x=col, kde=True, ax=axes[0])
  axes[0].set_title(f"Distribution of col {col}")
  sns.boxplot(data=df, x=col, ax=axes[1])
  axes[1].set_title(f"Boxplot of {col}")
  plt.tight_layout()
  plt.show()
```

```python
for col in columns:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1

  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr

  print(f"\n{col}")
  print(df[col].skew(), 2)
  print("IQR:", round(iqr, 2))
  print("Potential Outliers:", ((df[col] < lower) | (df[col] > upper)).sum())
```

```python
col = "Annual Income (k$)"

q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

mask = (df[col] < lower) | (df[col] > upper)

outliers = df.loc[mask]

print(outliers)
```

```python
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)

print(X.shape)
print(x_scaled.shape)
print(x_scaled.mean(axis=0))
print(x_scaled.std(axis=0))
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1,11):
   kmean = KMeans(n_clusters=i, init="k-means++", random_state=42)
   kmean.fit(x_scaled)
   wcss.append(kmean.inertia_)
plt.plot(range(1,11), wcss)
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Number of clusters count graph")
plt.show()
```

```python
k_mean_model = KMeans(n_clusters = 5, init="k-means++", random_state= 42)
labels = k_mean_model.fit_predict(x_scaled)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 11):
    model = KMeans(
        n_clusters=k,
        init="k-means++",
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(x_scaled)
    score = silhouette_score(x_scaled, labels)

    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

kmeans = KMeans(
    n_clusters=5,
    init="k-means++",
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(x_scaled)

centers = scaler.inverse_transform(kmeans.cluster_centers_)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="tab10",
    s=80,
    hue_order=sorted(df["Cluster"].unique())
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker="X",
    s=250,
    color="black",
    label="Centroids"
)

plt.title("Customer Segmentation Using K-Means (K=5)")
plt.legend()
plt.show()
```

```python
print(df.groupby("Cluster")[
    ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
].mean().round(2))

print("\nCustomers per cluster:")
print(df["Cluster"].value_counts().sort_index())
```
